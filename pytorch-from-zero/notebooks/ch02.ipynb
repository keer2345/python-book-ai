### Softmax 函数

````python
import numpy as np

def softmax(x):
    x = np.asarray(x, dtype=np.float64)
    if x.ndim == 1:
        x = x - np.max(x)
        e = np.exp(x)
        return e / np.sum(e)
    # 二维：按行做 softmax
    x = x - np.max(x, axis=1, keepdims=True)
    e = np.exp(x)
    return e / np.sum(e, axis=1, keepdims=True)
````

````python
a = np.array([[1,2,1,2,1,1,3]])
print(softmax(a))
````
